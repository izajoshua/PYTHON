Stacked barplot to know the churn rate when the customer is considered as a senior citizen in percentage to underdtand the data and graph more better 

```python
# Create the crosstab (counts)


ct = pd.crosstab(df["SeniorCitizen"], df["Churn"])

# Convert to % (row-wise)
ct_percent = ct.div(ct.sum(axis=1), axis=0) * 100

# Plot
plt.figure(figsize=(6,4))

# Bottom bar starts at 0
bottom = None

for churn_value in ct_percent.columns:
    plt.bar(
        ct_percent.index,
        ct_percent[churn_value],
        bottom=bottom,
        label=churn_value
    )

    # Add percentage labels
    for i, val in enumerate(ct_percent[churn_value]):
        plt.text(
            i,
            (bottom[i] if bottom is not None else 0) + val/2,
            f"{val:.1f}%",
            ha="center", va="center", fontsize=10, color="white"
        )

    # Update bottom for stacking
    bottom = ct_percent[churn_value] if bottom is None else bottom + ct_percent[churn_value]

plt.title("Stacked Bar Chart: Senior Citizen vs Churn (%)")
plt.ylabel("Percentage")
plt.xticks([0, 1], ["No", "Yes"])

plt.legend(title="Churn")
plt.show()
```

Comparitively a greter percentange of people in senior citizen catogory have churned out

```python
# List of categorical columns
cat_cols = ['gender', 'SeniorCitizen', 'Partner', 'Dependents',
            'PhoneService', 'MultipleLines', 'InternetService',
            'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
            'TechSupport', 'StreamingTV', 'StreamingMovies', 'Contract',
            'PaperlessBilling', 'PaymentMethod', 'Churn', 'SeniorCitizenLabel']

for col in cat_cols:
    plt.figure(figsize=(5,3))
    ax = sns.countplot(data=df, x=col, hue="Churn")
    plt.title(f"Countplot for {col}")
    
    # Add labels only for the main bars (not hue)
    if len(ax.containers) > 0:
        for container in ax.containers:
            ax.bar_label(container, padding=2)
    
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

The above visulisations present count plots showing customer churn(Yes or No) across various services features such as PhoneService, Multiplelines, Interservice, and OnlinSecurity.The majority of customers who do not churn tend to have services like PhoneServices, internetservices(Particulary DSL), and Online securoty enabled. For Services like online backup, Techsupport, and streamingTV, churn rates are noticebly higher when these services are not used or are unavailable. The charts emphasizes the rekatioship betwenn service usage and customer churn
